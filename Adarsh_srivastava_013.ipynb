```python
from transformers import pipeline
import pandas as pd

sentiment_pipeline = pipeline('sentiment-analysis', model='distilbert-base-uncased-finetuned-sst-2-english')
print("Model loaded successfully!")
```

```python
texts = [
    "I love this product, it's absolutely fantastic!",
    "This movie was terrible, a complete waste of time.",
    "The weather today is neither good nor bad.",
    "I am so happy and excited about the upcoming holidays.",
    "This is the worst experience I've ever had."
]

sentiment_results = sentiment_pipeline(texts)

df_sentiment = pd.DataFrame(sentiment_results)
df_sentiment['text'] = texts
display(df_sentiment[['text', 'label', 'score']])
```
